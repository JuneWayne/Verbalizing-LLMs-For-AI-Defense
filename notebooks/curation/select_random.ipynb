## How the pairs are selected

Each eligible pair has the same chance of being selected, and the fixed seed makes this random sample reproducible. The function `selection()` reads those 256 pairs into the table variable `selected`, then I count their injection methods and display the first ten selection records.

For random selection, the column `selection_score` is the random number used to order the candidates.


```python
# compare the 256 pairs used to train this method's adapter
selected = selection("random")
# the counts show which injection methods appear in this training group
display(selected.groupby("attack_method").size().rename("selected pairs").to_frame())
display(selected[["pair_id", "selection_score", "witness_redirection"]].head(10))
```
